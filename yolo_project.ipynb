```python
import os
from glob import glob

class_map = {
    'open_palm': 0,
    'close_palm': 1,
    'fist': 2,
    'four_fingers': 3,
    'next': 4,
    'prev': 5
}

base_label_dir = r"رابط بيانات ال train"

for class_name, correct_index in class_map.items():
    label_folder = os.path.join(base_label_dir, class_name)
    if not os.path.exists(label_folder):
        print(f" Folder {label_folder} not found, skipping.")
        continue

    label_files = glob(os.path.join(label_folder, '*.txt'))

    for file_path in label_files:
        with open(file_path, 'r') as f:
            lines = f.readlines()

        new_lines = []
        for line in lines:
            parts = line.strip().split()
            if len(parts) == 5:
                parts[0] = str(correct_index)
                new_lines.append(' '.join(parts))

        with open(file_path, 'w') as f:
            f.write('\n'.join(new_lines))

        print(f"✅ Updated: {file_path}")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data=r"رابط ال yaml",
    epochs=20,  # you can increase this if early stopping is enabled
    imgsz=640,
    device='cpu',
    patience=5  # Stop if no improvement in val metrics for 5 epochs
)
```

```python
from ultralytics import YOLO
import cv2

# Load the trained model
model = YOLO(".pt")

# Open webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLOv8 inference on the frame
    results = model.predict(source=frame, save=False, show=True, conf=0.3)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
